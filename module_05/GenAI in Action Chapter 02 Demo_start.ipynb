> **Note for this semester (2026):** the recorded demo video was made with last year's version of this notebook.
> Two things have changed since then, so your numbers will not match the video exactly:
>
> - **Token encoding:** this notebook uses `o200k_base` instead of `cl100k_base`. `o200k_base` is the tokenizer of
>   current OpenAI models (GPT-4o and the GPT-5 family); `cl100k_base` belongs to GPT-3.5 and GPT-4. Token IDs and
>   counts differ a little, and text in other languages usually takes fewer tokens.
> - **Embedding model:** `get_embedding()` uses `text-embedding-3-small` instead of the older `text-embedding-ada-002`.
>   It still returns 1,536 numbers, but the values are different.
>
> The steps and ideas in the video are the same.

```python
import tiktoken as tk

def count_tokens(string: str, encoding_name: str) -> int:
    # Get the encoding
    encoding = tk.get_encoding(encoding_name)
    # Encode the string
    encoded_string = encoding.encode(string)
    # Count the number of tokens
    num_tokens = len(encoded_string)
    return num_tokens

# Define the input string
prompt = "I have a white dog named Champ."

# Display the number of tokens in the String
print("Number of tokens:" , count_tokens(prompt, "o200k_base"))
```

## Get tokens and return the original string based on tokens

```python
import tiktoken as tk

def get_tokens(string: str, encoding_name: str) -> str:
    # Get the encoding
    encoding = tk.get_encoding(encoding_name)
    # Encode the string
    return encoding.encode(string)

def get_string(tokens: str, encoding_name: str) -> str:
    # Get the encoding
    encoding = tk.get_encoding(encoding_name)
    # Decode the tokens
    return encoding.decode(tokens)

# Define the input string
prompt = "I have a white dog named Champ."

print("o200k_base Tokens:" , get_tokens(prompt, "o200k_base"))
print("cl100k_base Tokens:" , get_tokens(prompt, "cl100k_base"))   # the encoding used in the video
print("p50k_base Tokens:" , get_tokens(prompt, "p50k_base"))
print("r50k_base Tokens:" , get_tokens(prompt, "r50k_base"))
```

```python
print("Original String:" , get_string([40, 679, 261, 6461, 6446, 11484, 15167, 13], "o200k_base"))
```

## Getting an embedding Using OpenAI embedding model

[OpenAI Embedding Models](https://platform.openai.com/docs/guides/embeddings#embedding-models)

```python
from dotenv import load_dotenv
from openai import OpenAI
import os

# define the environment path. it is at the parent folder
env_path ="../.env"

load_dotenv(dotenv_path=env_path, override=True)

api_key = os.getenv("OPENAI_API_KEY")

client = OpenAI()

def get_embedding(text):
    response = client.embeddings.create(
        model="text-embedding-3-small", input=text)
    return response.data[0].embedding
```

```python
embeddings = get_embedding("I have a white dog named Champ.")
print("Embedding Length:", len(embeddings))
print("Embedding:", embeddings[:5])
```
